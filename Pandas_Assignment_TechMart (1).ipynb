# Pandas Assignment - TechMart Sales Dataset

Load `TechMart_Sales_300Rows.csv` and solve the following questions.

## Q1. Load the CSV file.

```python

import pandas as pd
df=pd.read_csv('TechMart_Sales_300Rows.csv')
df
```

```python

df.isnull().sum()
```

## Q6. Fill missing Rating with 0.

```python

df['Rating']=df['Rating'].fillna(0)
df['Rating']
```

## Q7. Rename `Customer_Name` to `Customer`.

```python

df=df.rename(columns={'Customer_Name':'Customer'})
print(df.columns)
```

## Q8. Delete `Brand` column.

```python

df=df.drop('Brand',axis= 1)
print(df.columns)
```

## Q9. Create `Total_Amount = Price * Quantity`.

```python

df['Total_Amount']=df['Price']*df['Quantity']
df['Total_Amount']
```

```python

df[['Customer','Product','Total_Amount']]
```
